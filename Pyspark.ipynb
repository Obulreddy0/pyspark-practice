```python
from pyspark.sql import SparkSession

# 1. Create SparkSession
spark = SparkSession.builder \
    .appName("Day1_Basics") \
    .master("local[*]") \
    .getOrCreate()

# 2. Create DataFrame from Python list
data = [
    (1, "Rahul", "Engineering", 75000),
    (2, "Priya", "Marketing", 65000),
    (3, "Amit", "Engineering", 80000),
]
columns = ["emp_id", "name", "department", "salary"]

df = spark.createDataFrame(data, columns)

# 3. Actions — these TRIGGER execution
df.show()           # Display the table
df.printSchema()    # Show column names + types
print(df.count())   # Count rows

# 4. Always stop when done
#spark.stop()
```
