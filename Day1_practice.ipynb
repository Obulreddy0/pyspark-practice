```python
# Read all 4 CSV files (customers, orders, transactions, products) into separate DataFrames.
# For each, print the file name, number of columns, number of rows, and column names.

customers_df = spark.read.option("header", "true").option("inferSchema", "true").csv("pyspark-practice/data/customers.csv")
orders_df = spark.read.option("header", "true").option("inferSchema", "true").csv("pyspark-practice/data/orders.csv")
transactions_df = spark.read.option("header", "true").option("inferSchema", "true").csv("pyspark-practice/data/transactions.csv")
products_df = spark.read.option("header", "true").option("inferSchema", "true").csv("pyspark-practice/data/products.csv")

def analyze_dataframe(df, df_name):
    print(f"\n--- {df_name} ---")
    print(f"Number of columns: {len(df.columns)}")
    print(f"Number of rows: {df.count()}")
    print(f"Column names: {df.columns}")

analyze_dataframe(customers_df, "customers.csv")
analyze_dataframe(orders_df, "orders.csv")
analyze_dataframe(transactions_df, "transactions.csv")
analyze_dataframe(products_df, "products.csv")
```

Create a DataFrame for bank accounts using a manual schema (StructType) — not inferSchema. Columns:

Column	    Type

account_id	 String

holder_name	 String

balance	 Double

is_active	 Boolean

opened_date	 String

Add 4 rows and verify with .printSchema().

```python
from pyspark.sql.types import StructType, StructField, StringType, DoubleType, BooleanType

# Define schema manually
schema = StructType([
    StructField("account_id", StringType(), True),
    StructField("holder_name", StringType(), True),
    StructField("balance", DoubleType(), True),
    StructField("is_active", BooleanType(), True),
    StructField("opened_date", StringType(), True),
])

data = [
    ("ACC001", "Rahul Sharma", 45000.50, True, "2023-01-15"),
    ("ACC002", "Priya Patel", 120000.00, True, "2022-06-20"),
    ("ACC003", "Amit Kumar", 8500.75, False, "2021-11-10"),
    ("ACC004", "Sneha Reddy", 250000.00, True, "2023-03-05"),
]

df = spark.createDataFrame(data, schema)
df.show()
df.printSchema()
```

Exercise 7: Prove lazy evaluation. Read orders.csv, apply a filter (status == 'delivered'), add print statements before and after .show(). Question: When does the filter actually execute and why?

```python
orders_df = spark.read.option("header", "true").option("inferSchema", "true") \
    .csv("pyspark-practice/data/orders.csv")

# Transformation — LAZY, nothing runs here
filtered_df = orders_df.filter(orders_df["status"] == "delivered")

print("✅ Before show — filter is defined but NOT executed yet")

# Action — THIS triggers execution of everything above
filtered_df.show()

print("✅ After show — filter was executed during .show()")
```

Exercise 8:
SparkSession singleton. Create a SparkSession with appName("App_A"), then another with appName("App_B") using .getOrCreate(). Print both app names. Question: Are they the same object or different? Why?

Answer: SparkSession is a Singleton. .getOrCreate() means:

If a session already exists → return the existing one (ignores new appName)
If no session exists → create a new one


```python
spark_a = SparkSession.builder.appName("App_A").master("local[*]").getOrCreate()
print(f"Session A app name: {spark_a.sparkContext.appName}")

spark_b = SparkSession.builder.appName("App_B").master("local[*]").getOrCreate()
print(f"Session B app name: {spark_b.sparkContext.appName}")

print(f"Same object? {spark_a is spark_b}")
```
